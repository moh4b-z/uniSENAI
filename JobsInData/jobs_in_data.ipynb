## Colunas:

- <span style="color:#11c83f">work_year:</span>  O ano em que os dados foram registrados. Este campo indica o contexto temporal dos dados, importante para a compreensão das tendências salariais ao longo do tempo.

- <span style="color:#11c4c8">job_title:</span>  O título específico da função, como 'Cientista de Dados', 'Engenheiro de Dados' ou 'Analista de Dados'. Esta coluna é crucial para entender a distribuição salarial entre as diversas funções especializadas na área de dados.

- <span style="color:#11c83f">job_category:</span>  Uma classificação da função em categorias mais amplas para facilitar a análise. Isso pode incluir áreas como 'Análise de Dados', 'Aprendizado de Máquina', 'Engenharia de Dados', etc.

- <span style="color:#11c4c8">salary_currency:</span>  A moeda em que o salário é pago, como USD, EUR, etc. Isso é importante para a conversão de moeda e para entender o valor real do salário em um contexto global.

- <span style="color:#11c83f">salary:</span>  O salário bruto anual do cargo na moeda local. Este valor bruto do salário é fundamental para comparações salariais regionais diretas.

- <span style="color:#11c4c8">salary_in_usd:</span>  O salário bruto anual convertido para dólares americanos (USD). Essa conversão de moeda uniforme facilita comparações e análises salariais globais.

- <span style="color:#11c83f">employee_residence:</span>  O país de residência do funcionário. Este dado pode ser usado para explorar diferenças salariais geográficas e variações no custo de vida.

- <span style="color:#11c4c8">experience_level:</span>  Classifica o nível de experiência profissional do funcionário. As categorias comuns podem incluir "Iniciante", "Pleno", "Sênior" e "Executivo", fornecendo informações sobre como a experiência influencia o salário em funções relacionadas a dados.

- <span style="color:#11c83f">employment_type:</span>  Especifica o tipo de vínculo empregatício, como 'Tempo integral', 'Tempo parcial', 'Contrato', etc. Isso auxilia na análise de como diferentes modalidades de emprego afetam as estruturas salariais.

- <span style="color:#11c4c8">work_setting:</span>  O ambiente ou modalidade de trabalho, como 'Remoto', 'Presencial' ou 'Híbrido'. Esta coluna reflete o impacto dos ambientes de trabalho nos níveis salariais no setor de dados.

- <span style="color:#11c83f">company_location:</span>  O país onde a empresa está localizada. Isso ajuda a analisar como a localização da empresa afeta as estruturas salariais.

- <span style="color:#11c4c8">company_size:</span>  O tamanho da empresa empregadora, geralmente categorizado em pequeno (P), médio (M) e grande (G). Isso permite analisar como o tamanho da empresa influencia o salário.

## DataFrame em Python

Uma **estrutura de dados tabular** é uma forma de organizar informações em formato de **tabela**, composta por:

- **Linhas** (registros)
- **Colunas** (atributos ou variáveis)

Esse formato é semelhante a uma planilha do Excel ou a uma tabela de banco de dados.

Exemplo:

| Nome     | Idade | Profissão        |
|----------|--------|-----------------|
| Ana      | 25     | Cientista de Dados |
| Carlos   | 30     | Engenheiro      |

Cada linha representa um registro, e cada coluna representa uma característica desse registro.

---

Um **DataFrame** é uma estrutura de dados **bidimensional e tabular** fornecida pela biblioteca **Pandas** no Python.

Ele permite:

- Armazenar dados em formato de tabela
- Manipular dados (filtrar, ordenar, agrupar)
- Realizar análises estatísticas
- Limpar e transformar dados

O DataFrame é uma das estruturas mais importantes para análise de dados em Python.

---

# Tipos de Variáveis em Ciência de Dados

Em análise de dados, as variáveis são classificadas de acordo com o tipo de informação que representam. Elas podem ser **Qualitativas (Categóricas)** ou **Quantitativas (Numéricas)**.

---

### 1. Variáveis Qualitativas (Categóricas)

São variáveis que representam **características ou categorias**, e não números com significado matemático.

Elas se dividem em:

### Nominais

- Não possuem ordem natural.
- Servem apenas para classificação.
Mesmo que sejam representadas por números, esses números não indicam quantidade.

---

## Ordinais

- Possuem uma **ordem natural**.
- Indicam hierarquia ou nível.

---

# 2. Variáveis Quantitativas (Numéricas)

São variáveis que representam **valores numéricos** e permitem cálculos matemáticos.

Elas se dividem em:

## Discretas

- Assumem valores inteiros.
- Geralmente representam contagens.

## Contínuas

- Podem assumir qualquer valor dentro de um intervalo.
- Podem conter casas decimais.


| Tipo | Subtipo | Característica |
|------|---------|---------------|
| Qualitativa | Nominal | Sem ordem |
| Qualitativa | Ordinal | Com ordem |
| Quantitativa | Discreta | Valores inteiros |
| Quantitativa | Contínua | Valores decimais |

# 📂 Informações Faltantes e Duplicadas em um Banco de Dados

Em Ciência de Dados, antes de realizar qualquer análise, é fundamental verificar a **qualidade dos dados**. Dois problemas muito comuns são:

- Valores nulos (informações faltantes)
- Dados duplicados

Esses fatores podem comprometer análises, gerar resultados incorretos e prejudicar modelos de Machine Learning.

---

# 🔹 1. Valores Nulos (Informações Faltantes)

## 📌 O que são valores nulos?

Valores nulos representam **dados ausentes ou não informados** em uma base de dados.

No Pandas, eles geralmente aparecem como:

- `NaN` (Not a Number)
- `None`

### 🔎 Exemplo:

| Nome   | Idade | Salário |
|--------|--------|---------|
| Ana    | 25     | 3000    |
| Carlos | NaN    | 4000    |
| Maria  | 28     | NaN     |

---

## ⚠ Por que analisar valores nulos?

Porque eles podem:

- Distorcer médias e estatísticas
- Causar erros em cálculos
- Prejudicar modelos de Machine Learning
- Indicar falhas na coleta de dados

Antes da análise, é necessário decidir se os valores nulos serão:

- Removidos
- Substituídos (por média, mediana, moda, etc.)
- Mantidos (dependendo do caso)

---

# 🔹 2. Dados Duplicados

## 📌 O que são dados duplicados?

São registros ou informações repetidas dentro do banco de dados.

### 🔎 Exemplo:

| Nome | Idade | Cargo |
|------|--------|--------|
| João | 30     | Analista |
| João | 30     | Analista |

---

## ⚠ Por que analisar duplicatas?

Dados duplicados podem:

- Inflar resultados estatísticos
- Gerar contagens incorretas
- Comprometer análises de frequência
- Prejudicar modelos preditivos

Por isso, é essencial verificar e tratar duplicações antes de qualquer análise.

---

##### Executando o Código

````python
df.isnull().sum()
df.T.duplicated()
````
