- 22. Leia uma lista de N números (N informado pelo usuário) e calcule a média apenas dos
números entre 0 e 100 (ignore os demais).

```python
N = [10, 20, 30, 40, 500, -60, 60, 70, 80, 90, 10]
soma = 0
contador = 0
for n in N:
    if n >= 0 or n <= 100:
        soma += n
        contador += 1
print("Média:", soma / contador)
```

- 23. Leia N números e determine o segundo maior valor (considere que pode haver repetidos).

```python
Numeros = [10, 20, 30, 40, -60, 60, 60,70, 80, 90, 10]
maior = Numeros[0]
segundo_maior = Numeros[0]
for n in Numeros:
    if n >= maior:
        segundo_maior = maior
        maior = n

print("O segundo maior é", segundo_maior)
```

- 4. Pesquisa Linear (retornar índice)
    - Enunciado: Crie uma função que faça pesquisa linear em uma lista e retorne o índice do elemento se encontrado ou -1 se não estiver presente.

```python
def pesquisaLinear(lista, valor):
    for i in range(len(lista)):
        if lista[i] == valor:
            return i
    return -1

print(pesquisaLinear(listaV, 100))
```

- 5. Exclusão de Elemento (sem usar remove)
    - Enunciado: Crie uma função que remova a primeira ocorrência de um elemento de uma lista sem utilizar o método remove.

```python
def exclusaoSemRemove(lista, valor):
    for i in range(len(lista)):
        if lista[i] == valor:
            for j in range(i, len(lista) - 1):
                lista[j] = lista[j + 1]
            lista.pop()
            return lista
    return lista

print(exclusaoSemRemove(listaV, 10))
```

- 6. Inserção em Lista Ordenada
    - Enunciado: Crie uma função que insira um elemento em uma lista ordenada, mantendo a ordem sem usar sort.

```python
listaTeste = [1, 2, 3, 4, 5]

def insercaoOrdenada(lista, valor):
    lista.append(valor)
    for contador in range(len(lista) - 1):
        for contador2 in range(contador + 1, len(lista)):
            if lista[contador] > lista[contador2]:
                lista[contador], lista[contador2] = lista[contador2], lista[contador]
    return lista

print(insercaoOrdenada(listaTeste, 0))
```

    - Enunciado: Crie uma função que receba duas listas de mesmo tamanho (vetores) e retorne uma nova lista com a soma elemento a elemento.

```python
def listaDuplicada(lista):
	listaD = []
	quantidade = len(lista)
	comeco = 0
	while comeco <= quantidade:
		valor = lista[comeco]
		for i in range(comeco + 1, quantidade + 1):
            if lista[i] == valor:
			    listaD.append(lista[i])
				break
        comeco += 1
	return listaD
```
