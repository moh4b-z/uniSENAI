```python
avore = {}

valores = [57, 68, 97, 82, 46, 30, 90, 100, 75, 80, 85, 92, 88, 91, 94, 89, 93, 95, 96, 98]

for valor in valores:
    if len(avore) == 0:
        avore[valor] = {"esquerda": None, "direita": None}
        raiz = valor
    else:
        atual = raiz
        while True:
            if valor < atual:
                if avore[atual]["esquerda"] is None:
                    avore[atual]["esquerda"] = valor
                    avore[valor] = {"esquerda": None, "direita": None}
                    break
                else:
                    atual = avore[atual]["esquerda"]
            else:
                if avore[atual]["direita"] is None:
                    avore[atual]["direita"] = valor
                    avore[valor] = {"esquerda": None, "direita": None}
                    break
                else:
                    atual = avore[atual]["direita"]
```
