25. Simule um caixa: peça o valor da compra e o valor pago. Se o pago for menor, informe saldo
devedor. Se for maior, calcule o troco e informe quantas notas de 100, 50, 20, 10, 5, 2 e
moedas de 1 são necessárias (use for/if para decompor).

```python
def caixaDeMercado(valorCompra, valorPago):
    if valorCompra > valorPago:
        return "Valor pago é insuficiente. Faltam R$" + str(valorCompra - valorPago)
    elif valorCompra < valorPago:
        troco = valorPago - valorCompra
        notas = [100, 50, 20, 10, 5, 2, 1]
        resultado = []
        for nota in notas:
            qtd_notas = troco // nota
            if qtd_notas > 0:
                resultado.append((nota, qtd_notas))
            troco %= nota
        return "Valor pago é maior que o necessário. Troco: R$" + str(valorPago - valorCompra) + ". Notas e moedas: " + str(resultado)
    else:
        return "Valor pago é exato. Sem troco."

caixaDeMercado(40, 50)
```

28. Leia um número N e verifique se ele é um número perfeito (soma dos divisores próprios =
N).

```python
def verificacaoNumeroPerfeito(n):
    if n < 1:
        return False
    soma_divisores = 0
    for i in range(1, n):
        if n % i == 0:
            soma_divisores += i
    return soma_divisores == n

verificacaoNumeroPerfeito(6)
```

30. Peça um número N e imprima os N primeiros termos da sequência de Fibonacci (começando
em 0, 1).

```python
def fibonacci(n):
    n = n + 2
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    elif n == 2:
        return [0, 1]
    else:
        fib_sequence = [0, 1]
        for i in range(2, n):
            next_fib = fib_sequence[i-1] + fib_sequence[i-2]
            fib_sequence.append(next_fib)
        return fib_sequence

fibonacci(2)
```

- 27. Ordenação sem Sort (Selection Sort)
    - Enunciado: Crie uma função que ordene uma lista em ordem crescente implementando o algoritmo de Selection Sort, sem usar métodos nativos de ordenação.

```python
def ordenarLista(lista):
    if len(lista) <= 1:
        return lista
    else:
        pivo = lista[0]
        menores = [x for x in lista[1:] if x < pivo]
        maiores = [x for x in lista[1:] if x >= pivo]
        return ordenarLista(menores) + [pivo] + ordenarLista(maiores)
    
print(ordenarLista(listaV3))
```

```python
def contar_vogais(texto):
    vogais = "aeiouAEIOU"
    count = 0
    for char in texto:
        if char in vogais:
            count += 1
    return count

print(contar_vogais("Hello, World!"))
```

soma de todos os elementos.

```python
def soma_lista(lista):
    soma = 0
    for n in lista:
        soma += n
    return soma
print(soma_lista(listaV3))
```

8) Crie uma função chamada multiplicar_lista(lista) que receba uma lista de números e
retorne o produto (multiplicação) de todos os elementos.

```python
def multiplicar_lista(lista):
    produto = 1
    for n in lista:
        produto *= n
    return produto

print(multiplicar_lista(listaV3))
```

9) Crie uma função chamada eh_multiplo(a, b) que retorne True se 'a' for múltiplo de 'b' e
False caso contrário.

```python
def eh_multiplo(a, b):
    return a % b == 0

print(eh_multiplo(10, 2))
```

10) Crie uma função chamada contador_regressivo(n) que imprima uma contagem
regressiva de n até 0 usando um laço de repetição.

```python
def contador_regressivo(n):
    for i in range(n, -1, -1):
        print(i)

contador_regressivo(5)
```
