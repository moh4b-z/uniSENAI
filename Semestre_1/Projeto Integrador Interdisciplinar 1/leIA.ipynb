```python
import pandas as pd
# Alterado para 'leia' em minúsculo conforme a biblioteca instalada
import sys
import subprocess

subprocess.check_call([sys.executable, "-m", "pip", "install", "leia-br"])
from leia import SentimentIntensityAnalyzer 

# Inicializa o analisador
analyzer = SentimentIntensityAnalyzer()

def classificar_sentimento_numerico(score):
    if score <= -0.6:
        return 1 # Muito Negativo
    elif -0.6 < score <= -0.2:
        return 2 # Negativo
    elif -0.2 < score <= 0.2:
        return 3 # Neutro
    elif 0.2 < score <= 0.6:
        return 4 # Positivo
    else:
        return 5 # Muito Positivo

def aplicar_leia_sentimento(df, nome_coluna_texto):
    if (nome_coluna_texto in df.columns):
        print("Analisando sentimentos...")
        
        # O LeIA retorna um dicionário, pegamos apenas o 'compound'
        df['sentimento_score'] = df[nome_coluna_texto].apply(
            lambda x: analyzer.polarity_scores(str(x))['compound'] if pd.notnull(x) else 0.0
        )
        
        # Aplica a classificação inteira
        df['sentimento_nota'] = df['sentimento_score'].apply(classificar_sentimento_numerico)
    else:
        print(f"Erro: Coluna '{nome_coluna_texto}' não encontrada.")
    return df

# Exemplo de uso integrando com seu arquivo:
df = pd.read_csv('./Banco_de_Dados_PII3_AWS/avaliacoes.csv')
df = aplicar_leia_sentimento(df, 'review_comment_message')
# df.to_csv('./Banco_de_Dados_PII3_AWS/avaliacoes_analisadas.csv', index=False)
```
