## 1. Gráfico de linhas
##### Sintaxe básica:
``` Python
    plt.plot(x, y)

````python
import matplotlib.pyplot as plt

x = [5,6,7,8,9]
y = [12,14,16,18,20]

plt.plot(x, y)
plt.title('Gráfico de linha - Simples')

plt.xlabel('Eixo da abscissas')
plt.ylabel('Eixo da ordenadas')

plt.show()
````

````python
import numpy as np

x = np.arange(0,10,0.02)
y = np.sin(x)

plt.plot(x,y)
plt.title('Gráfico da função Seno')

plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.show
````

#### Gráfico de linhas (adicionando elementos)

````python
plt.figure(figsize=(12,6), dpi = 300)
plt.plot(x,y, color='purple', linewidth=15)
plt.title('Gráfico da função Seno')

plt.xlabel('Eixo X', fontsize=15)
plt.ylabel('Eixo Y', fontsize=15)
plt.show
````

### 2. Gráficos Multiplos e Subplot
#### Gráficos em uma mesma figura
- Com a função subplots (), você pode desenhar vários gráficos em uma figura:
- Nesse primeiro exemplo, vamos plotar dois sub gráficos de linhas

````python
plt.figure(figsize=(12,6), dpi = 300)

plt.subplot(1,2,1)
plt.title('Gráfico 1')

plt.subplot(1,2,2)
plt.title('Gráfico 2')
plt.show
````

### Gráficos lado a lado

````python
plt.figure(figsize=(12,6), dpi = 300)

x = np.arange(0,10,0.1)
y = np.cos(x)

plt.subplot(1,2,1)
plt.plot(x,y, color='purple')


x = np.arange(0,10,0.1)
y = np.sin(x)

plt.subplot(1,2,2)
plt.plot(x,y)

plt.show
````

### Gráfico um em cima do outro

````python
plt.figure(figsize=(12,6), dpi = 300)
x = np.arange(0,10,0.1)
y = np.cos(x)

for i in range(1,5):
    plt.subplot(2,2,i)
    plt.plot(x,y, color='purple')

plt.show()
````

````python
plt.figure(figsize=(12,6), dpi = 300)

x = np.arange(0,10,0.1)
y = np.cos(x)

plt.subplot(2,2,1)
plt.plot(x,y, color='purple')


x = np.arange(0,10,0.1)
y = np.sin(x)

plt.subplot(2,2,2)
plt.plot(x,y)

x = np.arange(0,10,0.1)
y = np.tan(x)

plt.subplot(2,2,3)
plt.plot(x,y, color='red')


x = np.arange(0,10,0.1)
y = np.sinh(x)

plt.subplot(2,2,4)
plt.plot(x,y, color='coral')

plt.show()
````

### Múltiplos Gráficos com Subplot
- É possível plotar quantos gráficos quiser em uma figura, só precisar informar a quantidade de linhas e colunas que terá a figura, além do índice do gráfico.

````python
data = np.random.normal(0,10,10000)
plt.figure(figsize=(12,6))
plt.hist(data, bins=30, color='darkblue', edgecolor='black', alpha=0.5)
plt.title('distribuição')

plt.xlabel('Valores')
plt.ylabel('Frequência')
plt.show
````

````python
import seaborn as sns
data = np.random.normal(0,10,1000)
plt.figure(figsize=(12,6))
sns.histplot(data, bins=30, kde=True)
plt.title('distribuição')

plt.xlabel('Valores')
plt.ylabel('Frequência')
plt.show
````

### 4. Gráficos boxplot com Matplotlib e Seaborn
- Exemplos de boxplot, com a base de dados flores iris, que tem 3 tipos de flores

````python
df = sns.load_dataset('iris')
````

````python
plt.figure(figsize=(6,6))
sns.boxplot(x='species', y='petal_length', data=df, width=0.3, hue='species', palette='plasma')

plt.title('Distribution of Petal Length by Species')
plt.xlabel('Iris Species')
plt.ylabel('Petal Length (cm)')

plt.show()
````

### Mult Boxplot com Matplotlib

````python
setosa = df[df['species']=='setosa']['petal_length']
versicolor = df[df['species']=='versicolor']['petal_length']
virginica = df[df['species']=='virginica']['petal_length']

dados = [setosa, versicolor, virginica]
box = plt.boxplot(dados, patch_artist=True)
for patch, cor in zip(box['boxes'], ['#565656', '#5656f6', '#5656ff']):
    patch.set_facecolor(cor)
plt.title('Distribution of Petal Length by Species')
plt.xlabel('Iris Species')
plt.ylabel('Petal Length (cm)')

plt.show()
````

#### Formatando por partes

````python
plt.figure
box = plt.boxplot(dados, patch_artist=True, vert=False)

cores= ['pink', 'lightblue', 'lightgreen']

for i, bplot in enumerate(box['boxes']):
    bplot.set(color=cores[i], linewidth=3)

cores_w = ['pink', 'lightblue', 'lightgreen','gold', 'orchid', 'skyblue']
for i, whisker in enumerate(box['whiskers']):
    whisker.set(color=cores_w[i], linewidth=3)


for i, caps in enumerate(box['caps']):
    caps.set(color=cores_w[i], linewidth=3)

for i, medians in enumerate(box['medians']):
    medians.set(color=cores[i], linewidth=3)


plt.title('Distribution of Petal Length by Species')
plt.xlabel('Iris Species')
plt.ylabel('Petal Length (cm)')

plt.show()
````

### Mult Boxplot com Matplotlib Cor

````python
plt.figure
box = plt.boxplot(dados, notch=True, patch_artist=True)

cores= ['pink', 'lightblue', 'lightgreen']

for i, bplot in enumerate(box['boxes']):
    bplot.set(color='black', linewidth=1)
    bplot.set(facecolor= cores[i])

for i, whisker in enumerate(box['whiskers']):
    whisker.set(color='black', linewidth=1)


for i, caps in enumerate(box['caps']):
    caps.set(color='black', linewidth=1)

for i, medians in enumerate(box['medians']):
    medians.set(color='black', linewidth=1)

plt.title('Distribution of Petal Length by Species')
plt.xlabel('Iris Species')
plt.ylabel('Petal Length (cm)')

plt.show()
````

### Boxplot, Gráfico de Disperção com Seaborn

````python
plt.figure(figsize=(6,6))
ax = sns.boxplot(x='species', y='petal_length', data=df, width=0.3, hue='species', palette='plasma', linewidth=3)

# ax = sns.stripplot(x='species', y='petal_length', data=df, hue='species')

ax = sns.swarmplot(x='species', y='petal_length', data=df, hue='species')

plt.title('Distribution of Petal Length by Species')
plt.xlabel('Iris Species')
plt.ylabel('Petal Length (cm)')

plt.show()
````

### Traçar gráfico de violino com seaborn

````python
plt.figure(figsize=(6,6))
ax = sns.histplot(x='species', y='petal_length', data=df, hue='species', palette='plasma', linewidth=3)

plt.title('Distribution of Petal Length by Species')
plt.xlabel('Iris Species')
plt.ylabel('Petal Length (cm)')

plt.show()
````

#### Estimativa da densidade

````python
plt.figure(figsize=(6,6))
ax = sns.kdeplot(x='petal_length', data=df, hue='species', palette='plasma', linewidth=3)

plt.title('Distribution of Petal Length by Species')
plt.xlabel('Iris Species')
plt.ylabel('Petal Length (cm)')

plt.show()
````

#### Extensão do boxplot

````python
plt.figure(figsize=(6,6))
ax = sns.boxenplot(x='petal_length', data=df, hue='species', palette='plasma', linewidth=1)

plt.title('Distribution of Petal Length by Species')
plt.xlabel('Iris Species')
plt.ylabel('Petal Length (cm)')

plt.show()
````

### 5. Visualização de Dados Categóricos
#### Gráfico de Barras

````python
categorias = ['A', 'B','C','D','E']
valores = [43, 56, 87, 35, 90]
plt.figure(figsize=(12,6), dpi = 300)

plt.subplot(1,2,1)
plt.bar(categorias, valores, color=['r', 'b', 'k', 'g', 'y'])
plt.title('Gráfico 1')

plt.subplot(1,2,2)
plt.barh(categorias, valores, color=['r', 'b', 'k', 'g', 'y'])
plt.title('Gráfico 2')
plt.show
````

````python
import pandas as pd

df = pd.DataFrame({'categorias': categorias, 'valores': valores})

df
````

````python
plt.figure(figsize=(12,6), dpi = 300)
sns.barplot(x='categorias', y='valores', data=df, hue='categorias', palette = 'viridis')
plt.show()
````

````python
plt.figure(figsize=(12,6), dpi = 300)
plt.pie(valores, labels=categorias, autopct='%1.1f%%')
plt.show()
````

#### Gráfico de Torta ou Pizza

#### Plotly (interativo)

````python
import plotly.express as px

fig = px.pie(df, values='valores', names='categorias', title='Barra')
fig.update_layout(width=500, height=500)
fig.show()
````

### 6. Visualização de Relações entre Variáveis
#### Scatter Plot

````python
x = np.random.rand(100)

y = x + np.random.normal(0, 0.1, 100)
````

````python
plt.scatter(x,y, alpha=0.4)
plt.show()
````

````python
df = pd.DataFrame({'X':x, 'Y':y})
sns.jointplot(x = 'X', y = 'Y', data=df, kind='reg')
plt.tight_layout()
plt.show()
````
