## 1. O que é um dicionário?

Um **dicionário** é uma estrutura de dados que armazena informações no formato:

**chave : valor**

Diferente de listas e tuplas, que organizam dados por posição, o dicionário organiza dados por **associação**.

### Ideia central

Em vez de pensar em posições, no dicionário pensamos em rótulos, como:

- `"nome"`
- `"idade"`
- `"curso"`

Isso torna o dicionário muito útil para representar registros, cadastros, configurações, resultados e dados rotulados.

### Características principais

- é uma estrutura **mutável**;
- armazena pares `chave: valor`;
- o acesso é feito pela **chave**;
- não pode ter **duas chaves iguais**;
- pode armazenar diferentes tipos de dados;
- pode conter listas, tuplas e até outros dicionários.


## 2. Estrutura de um dicionário

A sintaxe básica de um dicionário usa **chaves `{}`**.

Exemplo geral:

```python
{
    "chave1": valor1,
    "chave2": valor2,
    "chave3": valor3
}
```

Cada item do dicionário possui:

- uma **chave**, que identifica o dado;
- um **valor**, que é a informação associada à chave.

## 5. Não pode ter duas chaves iguais

Em um dicionário, **as chaves devem ser únicas**.

Se uma chave for repetida, o valor mais recente sobrescreve o anterior.


````python
dados = {
    "nome": "Mohammad",
    "nome": "Salim",
    "idade": 12,
    "idade": 13,
    "idade": 18,
}
````

Observe que a chave `"nome"` apareceu duas vezes, mas o dicionário manteve apenas o último valor associado a ela.

Portanto:

- pode haver valores repetidos;
- **não pode haver duas chaves iguais** de forma útil, pois a última sobrescreve a anterior.
