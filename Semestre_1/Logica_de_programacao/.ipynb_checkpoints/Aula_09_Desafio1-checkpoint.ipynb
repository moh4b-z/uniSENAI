## Contexto

A proposta deste exercício não é discutir o jogo em si, mas usar esse contexto para resolver um problema de lógica de programação. O foco estará em modelar um processo produtivo simples, identificar dependências entre etapas e controlar o fluxo com estruturas de repetição e decisão.

## Situação-problema

Imagine uma base instalada em uma região extremamente fria. Para sobreviver e crescer, essa base precisa produzir equipamentos e vendê-los a um distribuidor externo.

O processo inicial funciona da seguinte maneira:

1. Um trabalhador coleta madeira.
2. Essa madeira é enviada para uma serraria.
3. A serraria transforma a madeira em madeira processada.
4. Uma oficina utiliza a madeira processada para fabricar equipamentos.
5. Esses equipamentos são colocados em um caminhão.
6. Quando o caminhão atinge sua capacidade máxima, ele parte para entrega.
7. Após a entrega, a base recebe moedas.
8. Enquanto o caminhão está ausente, a produção deve aguardar seu retorno.

## Objetivo pedagógico

- usar laços de repetição com critério de parada;
- usar estruturas condicionais para controlar o fluxo;
- empregar `continue` em situações em que uma etapa não pode prosseguir;
- acompanhar o estado de um sistema ao longo do tempo.

## Modelo lógico do sistema

Antes de programar, vale organizar a lógica do processo.

Em cada ciclo da simulação, o programa deve verificar:

1. Se o caminhão está disponível.
2. Se for o caso, produzir madeira.
3. Se houver madeira, processá-la.
4. Se houver madeira processada, fabricar um equipamento.
5. Se houver equipamento, carregá-lo no caminhão.
6. Se o caminhão estiver cheio, realizar a entrega.
7. Repetir até alcançar a meta de moedas.

Esse tipo de raciocínio é essencial em programação, porque ajuda a transformar um problema narrativo em uma sequência de regras executáveis.


## Configurações iniciais

Neste primeiro bloco vamos declarar as variáveis centrais do problema.

- `META_MOEDAS` define quando a simulação termina.
- `CAPACIDADE_CAMINHAO` define quantos equipamentos cabem no caminhão.
- `VALOR_POR_EQUIPAMENTO` define o retorno financeiro por item transportado.
- `madeira`, `madeira_processada` e `equipamentos` representam estoques intermediários.
- `carga_caminhao` guarda quantos equipamentos já foram colocados no transporte.
- `caminhao_disponivel` informa se a produção pode continuar normalmente.
- `ciclo` ajuda a visualizar o avanço da simulação.

Em problemas de lógica, a escolha correta das variáveis é parte importante da solução.


```python
META_MOEDAS = 100
MINHAS_MOEDAS = 0
madeiras_retiradas = 5
minhas_madeiras = 0
minhas_madeiras_processadas = 0
caminhao = {
    caminhao_disponivel = True,
    ciclos_espera: 3,
    ciclos_atual: 0,
    CAPACIDADE_CAMINHAO: 10,
    carga_caminhao: 0
}
madeira = {
    processada: False
}
equipamento = {
    madeiras_processadas_necessarias: 2,
    madeiras_processadas_atual: 0,
    preco: 3
}
ciclo = 3
```

## Laço principal da simulação

Agora vamos implementar a estrutura que fará o sistema funcionar repetidamente enquanto a meta ainda não for atingida.

**Dica:** Procure usar `while` para fazer os laços de repetição, pois não sabemos exatamente quantos ciclos serão necessários até atingir a meta. Isso torna o `while` mais adequado do que um `for` neste contexto.

```python

while MINHAS_MOEDAS < META_MOEDAS:
    minhas_madeiras += madeiras_retiradas
    print ('Quantidade de madeiras', minhas_madeiras)
    minhas_madeiras_processadas += minhas_madeiras
    minhas_madeiras = 0
    print ('Quantidade de madeiras processadas', minhas_madeiras)
    equipamentos_feitos = minhas_madeiras_processadas // equipamento.madeiras_processadas_necessarias
    minhas_madeiras_processadas = minhas_madeiras_processadas - equipamentos_feitos * equipamento.madeiras_processadas_necessarias
    print('equipamentos feitos', equipamentos_feitos)
    
```
