## 1. O que é um dicionário?

Um **dicionário** é uma estrutura de dados que armazena informações no formato:

**chave : valor**

Diferente de listas e tuplas, que organizam dados por posição, o dicionário organiza dados por **associação**.

### Ideia central

Em vez de pensar em posições, no dicionário pensamos em rótulos, como:

- `"nome"`
- `"idade"`
- `"curso"`

Isso torna o dicionário muito útil para representar registros, cadastros, configurações, resultados e dados rotulados.

### Características principais

- é uma estrutura **mutável**;
- armazena pares `chave: valor`;
- o acesso é feito pela **chave**;
- não pode ter **duas chaves iguais**;
- pode armazenar diferentes tipos de dados;
- pode conter listas, tuplas e até outros dicionários.


## 2. Estrutura de um dicionário

A sintaxe básica de um dicionário usa **chaves `{}`**.

Exemplo geral:

```python
{
    "chave1": valor1,
    "chave2": valor2,
    "chave3": valor3
}
```

Cada item do dicionário possui:

- uma **chave**, que identifica o dado;
- um **valor**, que é a informação associada à chave.

## 5. Não pode ter duas chaves iguais

Em um dicionário, **as chaves devem ser únicas**.

Se uma chave for repetida, o valor mais recente sobrescreve o anterior.


````python
dados = {
    "nome": "Mohammad",
    "nome": "Salim",
    "idade": 12,
    "idade": 13,
    "idade": 18,
}
````

Observe que a chave `"nome"` apareceu duas vezes, mas o dicionário manteve apenas o último valor associado a ela.

Portanto:

- pode haver valores repetidos;
- **não pode haver duas chaves iguais** de forma útil, pois a última sobrescreve a anterior.

## 11. For em dicionário

É muito comum percorrer dicionários com laços `for`.


### 11.1 Percorrendo apenas as chaves


````python
aluno = {
    "nome": "Mohammad",
    "idade": 18,
    "curso": "DS"
}

for i in aluno:
    print(i)
````

### 11.2 Percorrendo as chaves explicitamente com `keys()`


````python
for i in aluno.keys():
    print(i)
````

### 11.3 Percorrendo apenas os valores com `values()`


````python
for i in aluno.values():
    print(i)
````

### 11.4 Percorrendo chaves e valores com `items()`


````python
for i in aluno.items():
    print(i)
````

````python
for chave, values in aluno.items():
    print(i)
````

## 13. Colocando listas dentro de um dicionário

Um dicionário pode armazenar listas como valores.  
Isso é extremamente útil quando um mesmo campo precisa guardar vários elementos.


````python
aluno = {
    "nome": "Mohammad",
    "notas": [8.5, 9.2, 9.6],
    "disciplinas": ["Estrutura de dados", "Versionamento", "Lógica de progrmação"]
}
print(aluno)
````

````python
print(aluno['nome'])
print(aluno['notas'][2])
print(aluno['disciplinas'][2])
````

**Como calcular a média das notas armazenadas em uma lista dentro do dicionário?**

````python
media = sum(aluno["notas"]) / len(aluno["notas"])
print("Media", media)
````

## 15. Operações com dados armazenados em dicionários

Os valores armazenados em um dicionário podem ser usados em cálculos e operações.


### 15.1 Soma de valores


````python
vendas = {
    "Janeiro": 1200,
    "Fevereiro": 1500,
    "Marco": 1100,
}
````

### 15.2 Usando `sum()` com `values()`


````python
total = sum(vendas.values())
print("Total:",total)
````

### 15.3 Média de valores armazenados


````python
total = sum(vendas.values()) / len(vendas)
print("Total:",total)
````

### 15.4 Encontrando maior e menor valor


````python
print("Mes com maior venda", max(vendas.values()))
print("Mes com maior venda", min(vendas.values()))
````

### 15.5 Encontrando a chave associada ao maior valor

````python
mes_maior = max(vendas, key=vendas.get)
print("Mes com maior venda", mes_maior)
print("Mes com maior valor", vendas[mes_maior])
````

## 16. Atualizando vários itens ao mesmo tempo

O método `update()` permite inserir ou modificar vários pares chave-valor de uma vez.


````python
aluno = {
    "nome": "Mohammad",
    "idade": 18,
    "curso": "DS"
}
aluno.update({
    "estado": "SP",
    "idade": 19,
    "curso": "Ciencia de dados"
})

print(aluno)
````

## 17. Dicionários aninhados
