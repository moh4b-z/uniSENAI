# Significado das colunas

Dataset de **play-by-play** da NFL, temporadas **2009 a 2016** (`maxhorowitz/nflplaybyplay2009to2016`, pacote **nflscrapR**). Cada linha é um evento de uma partida (jogada, timeout, fim de período etc.).

As descrições seguem o dicionário do nflscrapR. Vários campos são indicadores **0/1** (ou `True`/`False` depois da conversão de tipo do gabarito): **1 / True** significa que o evento aconteceu.

## Jogo, tempo e situação de campo

| Coluna | Significado |
| --- | --- |
| `Date` | Data do jogo. |
| `GameID` | Identificador do jogo. |
| `Drive` | Número da série ofensiva (drive) dentro do jogo. Uma série vai da posse até o time perder a bola, pontuar ou o tempo acabar. |
| `qtr` | Quarto da partida: 1 a 4. O valor 5 é a prorrogação. |
| `down` | Descida da jogada: 1ª, 2ª, 3ª ou 4ª. Fica vazio em jogadas sem down (kickoff, ponto extra, timeout). |
| `time` | Tempo restante no quarto no início da jogada. |
| `TimeUnder` | Minutos restantes no quarto, arredondados para cima. Ex.: 14:53 vira 15. Vai de 1 a 15. |
| `TimeSecs` | Segundos restantes no jogo inteiro. |
| `PlayTimeDiff` | Diferença, em segundos, entre o relógio desta jogada e o da jogada anterior. |
| `SideofField` | Lado do campo em que está a linha de scrimmage (sigla do time dono desse lado). |
| `yrdln` | Linha de jardas, de 0 a 50 (a linha de 50 é o meio-campo). |
| `yrdline100` | Distância, em jardas, até a end zone do adversário. Vai de 1 a 99. Quanto menor, mais perto do touchdown. |
| `ydstogo` | Jardas que faltam para conseguir uma nova primeira descida (first down). |
| `ydsnet` | Jardas líquidas acumuladas na série até esta jogada. |
| `GoalToGo` | 1 se a jogada é **and goal**: a end zone está mais perto do que a marca do first down. |
| `posteam` | Time com a posse de bola (ataque). |
| `DefensiveTeam` | Time na defesa. No punt, quem recebe está na defesa; no kickoff, quem recebe está no ataque. |
| `HomeTeam` | Sigla do time mandante. |
| `AwayTeam` | Sigla do time visitante. |
| `Season` | Ano da temporada (2009 a 2016). Coluna criada ao juntar os arquivos anuais. |

## Resultado geral da jogada

| Coluna | Significado |
| --- | --- |
| `desc` | Texto livre com a narração oficial da jogada. É a fonte de onde várias colunas abaixo foram extraídas. |
| `PlayAttempted` | Marca a linha como uma jogada contabilizada. No dicionário original, deve valer 1. |
| `PlayType` | Tipo do evento. Valores possíveis: Kickoff, Punt, Pass, Sack, Run, Field Goal, Extra Point, Quarter End, Two Minute Warning, Half End, End of Game, No Play, QB Kneel, Spike, Timeout. |
| `Yards.Gained` | Jardas ganhas (ou perdidas, se negativo) nesta jogada. |
| `FirstDown` | 1 se a jogada converteu uma nova primeira descida. |
| `sp` | 1 se a jogada gerou pontuação de qualquer tipo (scoring play). |
| `Touchdown` | 1 se a jogada terminou em touchdown. |
| `Safety` | 1 se a jogada terminou em safety (2 pontos para a defesa). |
| `ExPointResult` | Resultado do ponto extra (chute de 1 ponto): Made, Missed, Blocked ou Aborted. |
| `TwoPointConv` | Resultado da conversão de 2 pontos do ataque: Success ou Failure. |
| `DefTwoPoint` | Resultado da conversão de 2 pontos da defesa (quando a defesa retorna a tentativa): Success ou Failure. |
| `Onsidekick` | 1 se o kickoff foi um onside kick (chute curto para tentar recuperar a bola). |
| `PuntResult` | Resultado do punt: Clean (normal) ou Blocked (bloqueado). |

## Passe

| Coluna | Significado |
| --- | --- |
| `Passer` | Nome do quarterback que lançou, quando a jogada foi um passe. |
| `Passer_ID` | Identificador GSIS da NFL desse passador. |
| `PassAttempt` | 1 se houve tentativa de passe. |
| `PassOutcome` | Resultado do passe: Complete ou Incomplete. |
| `PassLength` | Classificação da distância do passe: Short (curto) ou Deep (longo). |
| `AirYards` | Jardas que a bola percorreu no ar, em passes completos e incompletos. Negativo significa alvo atrás da linha de scrimmage. |
| `YardsAfterCatch` | Jardas ganhas pelo recebedor depois de pegar a bola (YAC). |
| `QBHit` | 1 se o quarterback foi derrubado no chão. |
| `PassLocation` | Lado do passe: left, middle ou right. |
| `InterceptionThrown` | 1 se o passe foi interceptado. |
| `Interceptor` | Nome do jogador que interceptou. |
| `Receiver` | Nome do recebedor alvo da jogada. |
| `Receiver_ID` | Identificador GSIS desse recebedor. |
| `Reception` | 1 se a bola foi recebida (passe completo). |
| `Sack` | 1 se o quarterback sofreu sack. |

## Corrida

| Coluna | Significado |
| --- | --- |
| `Rusher` | Nome do corredor, quando a jogada foi terrestre. |
| `Rusher_ID` | Identificador GSIS desse corredor. |
| `RushAttempt` | 1 se houve tentativa de corrida. |
| `RunLocation` | Lado da corrida: left, middle ou right. |
| `RunGap` | Buraco da linha usado na corrida: guard, tackle ou end. |

## Retorno, chute, fumble, revisão e penalidade

| Coluna | Significado |
| --- | --- |
| `ReturnResult` | Resultado do retorno de punt, kickoff, interceptação ou fumble: Fair Catch, Touchback ou Touchdown. |
| `Returner` | Jogador que retornou o punt ou o kickoff. |
| `BlockingPlayer` | Jogador que bloqueou o ponto extra, o field goal ou o punt. |
| `Tackler1` | Principal responsável pelo tackle. |
| `Tackler2` | Segundo jogador no tackle (assistência). |
| `FieldGoalResult` | Resultado do field goal: Good, No Good ou Blocked. |
| `FieldGoalDistance` | Distância do field goal, em jardas. |
| `Fumble` | 1 se houve fumble (bola perdida pelo portador). |
| `RecFumbTeam` | Time que recuperou o fumble. |
| `RecFumbPlayer` | Jogador que recuperou o fumble. |
| `Challenge.Replay` | 1 se a jogada foi para revisão de replay. |
| `ChalReplayResult` | Resultado da revisão: Upheld (lance mantido) ou Reversed (lance revertido). |
| `Accepted.Penalty` | 1 se houve penalidade aceita. |
| `PenalizedTeam` | Time penalizado. |
| `PenaltyType` | Nome da falta (ex.: False Start, Defensive Pass Interference, Holding, Roughing the Passer). |
| `PenalizedPlayer` | Jogador penalizado. |
| `Penalty.Yards` | Jardas marcadas por causa da penalidade. |

## Placar e timeouts

| Coluna | Significado |
| --- | --- |
| `PosTeamScore` | Placar do time com a posse (ataque). |
| `DefTeamScore` | Placar do time da defesa. |
| `ScoreDiff` | Diferença de placar: ataque menos defesa. Positivo = ataque na frente. |
| `AbsScoreDiff` | Módulo de `ScoreDiff`. Mostra o tamanho da diferença, sem dizer quem está na frente. |
| `Timeout_Indicator` | 1 se um timeout foi cobrado nesta linha. |
| `Timeout_Team` | Time que teve o timeout descontado. Fica vazio (ou None) quando não houve timeout. |
| `posteam_timeouts_pre` | Timeouts que o time com a posse ainda tinha no início da jogada. |
| `HomeTimeouts_Remaining_Pre` | Timeouts restantes do mandante no início da jogada. |
| `AwayTimeouts_Remaining_Pre` | Timeouts restantes do visitante no início da jogada. |
| `HomeTimeouts_Remaining_Post` | Timeouts restantes do mandante no fim da jogada. Também cai quando um challenge é perdido. |
| `AwayTimeouts_Remaining_Post` | Timeouts restantes do visitante no fim da jogada. Também cai quando um challenge é perdido. |

## Modelos: pontos esperados e probabilidade de vitória

Essas colunas não vêm da súmula. São saídas de modelos estatísticos do nflscrapR. "Próximo" significa o próximo evento de pontuação no mesmo tempo (half).

| Coluna | Significado |
| --- | --- |
| `No_Score_Prob` | Probabilidade de ninguém pontuar no restante do tempo. |
| `Opp_Field_Goal_Prob` | Probabilidade de a defesa marcar o próximo field goal. |
| `Opp_Safety_Prob` | Probabilidade de a defesa marcar o próximo safety. |
| `Opp_Touchdown_Prob` | Probabilidade de a defesa marcar o próximo touchdown. |
| `Field_Goal_Prob` | Probabilidade de o time com a posse marcar o próximo field goal. |
| `Safety_Prob` | Probabilidade de o time com a posse marcar o próximo safety. |
| `Touchdown_Prob` | Probabilidade de o time com a posse marcar o próximo touchdown. |
| `ExPoint_Prob` | Probabilidade de o time com a posse acertar o ponto extra. |
| `TwoPoint_Prob` | Probabilidade de o time com a posse converter os 2 pontos. |
| `ExpPts` | Pontos esperados (expected points) do time com a posse **antes** da jogada, dada a situação (down, distância, campo, tempo, placar). |
| `EPA` | Expected Points Added: quanto a jogada mudou os pontos esperados do time com a posse. Positivo = jogada boa para o ataque. |
| `airEPA` | Parte do EPA atribuída às jardas que a bola viajou no ar. |
| `yacEPA` | Parte do EPA atribuída às jardas depois da recepção. `airEPA` + `yacEPA` decompõem o EPA do passe. |
| `Home_WP_pre` | Probabilidade de vitória do mandante **antes** da jogada. |
| `Away_WP_pre` | Probabilidade de vitória do visitante **antes** da jogada. |
| `Home_WP_post` | Probabilidade de vitória do mandante **depois** da jogada. |
| `Away_WP_post` | Probabilidade de vitória do visitante **depois** da jogada. |
| `Win_Prob` | Probabilidade de vitória do time com a posse, no início da jogada. |
| `WPA` | Win Probability Added: quanto a jogada mudou a probabilidade de vitória do time com a posse. |
| `airWPA` | Parte do WPA atribuída às jardas no ar. |
| `yacWPA` | Parte do WPA atribuída às jardas depois da recepção. |
