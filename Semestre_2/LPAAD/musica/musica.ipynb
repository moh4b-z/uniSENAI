```python
class Musica:
    def __init__(self, titulo: str, cantor: str, duracao: int, genero: str, ano: int):

        if not isinstance(titulo, str) or not titulo.strip():
            raise ValueError("O título deve ser uma string não vazia.")
        if not isinstance(cantor, str) or not cantor.strip():
            raise ValueError("O cantor deve ser uma string não vazia.")
        if not isinstance(duracao, int) or duracao <= 0:
            raise ValueError("A duração deve ser um número inteiro positivo.")
        if not isinstance(genero, str) or not genero.strip():
            raise ValueError("O gênero deve ser uma string não vazia.")
        if not isinstance(ano, int) or ano <= 0:
            raise ValueError("O ano deve ser um número inteiro positivo.")

        self.titulo = titulo.strip()
        self.cantor = cantor.strip()
        self.duracao = duracao
        self.genero = genero.strip()
        self.ano = ano
        
        
        # Atributos de Controle
        self.reproducoes = 0
        self.tocando = False
        self.favorita = False

    @property
    def titulo(self):
        return self._titulo
    @property
    def cantor(self):
        return self._cantor
    @property
    def duracao(self):
        return self._duracao
    @property
    def genero(self):
        return self._genero
    @property
    def ano(self):
        return self._ano

    def reproduzir(self):
        if self.tocando == False:
            print(f"Reproduzindo música: {self._titulo}")
            self.reproducoes += 1
            self.tocando = True
        else:
            print("Esta música já esta sendo reproduzida")
            
    def pausar(self):
        if self.tocando == True:
            print(f"Pausando música: {self._titulo}")
            self.tocando = False
        else:
            print("A música não está sendo reproduzida.")
    
    def alterar_genero(self, genero):
        self._genero = genero

    def alterar_titulo(self, titulo):
        self.titulo = titulo

    def alterar_cantor(self, cantor):
        self._cantor = cantor

    def alterar_duracao(self, duracao):
        self._duracao = duracao

    def alterar_ano(self, ano):
        self._ano = ano

    def favoritar(self):
        self.favorita = True

    def desfavoritar(self):
        self.favorita = False

    def exibir_informacoes(self):  
        print(f"Música: {self._titulo}")
        print(f"Artista: {self._cantor}")
        print(f"Duração: {self._duracao} s")
        print(f"Reproduções: {self.reproducoes}")
        print(f"Gênero: {self._genero}")
        print(f"Ano: {self._ano}")
        print(f" favorita: {self.favorita}")
```

```python
m1 = Musica("Shape of You", "Ed Sheeran", 240, "Pop", 2017)

m1.exibir_informacoes()
```

```python
m1.alterar_ano(2018)
m1.alterar_duracao(250)
m1.alterar_genero("Pop Rock")
m1.alterar_titulo("Perfect")
m1.alterar_cantor("Ed Sheeran feat. Beyoncé")

m1.exibir_informacoes()
```
