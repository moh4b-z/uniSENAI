```python
from cryptography.fernet import Fernet
```

```python
chave = Fernet.generate_key()
print(f"Chave gerada (guarde em segredo): {chave.decode()}\n")

fernet = Fernet(chave)

mensagem_original = "Minha mensagem ultra secreta!".encode()
print(f"Mensagem original: {mensagem_original.decode()}")

mensagem_criptografada = fernet.encrypt(mensagem_original)
print(f"Mensagem criptografada (bytes): {mensagem_criptografada}\n")

mensagem_descriptografada = fernet.decrypt(mensagem_criptografada)
print(f"Mensagem descriptografada com sucesso: {mensagem_descriptografada.decode()}")
```

```python
from cryptography.hazmat.primitives.asymmetric import rsa, padding
from cryptography.hazmat.primitives import hashes
```

```python
chave_privada_gamma = rsa.generate_private_key(
    public_exponent=65537,
    key_size=2048
)
chave_publica_gamma = chave_privada_gamma.public_key()
```

```python
dados = "AAAAAAA BBBBB"

dados_cifrados = chave_publica_gamma.encrypt(dados.encode(),
    padding.OAEP(
        mgf=padding.MGF1(algorithm=hashes.SHA256()),
        algorithm=hashes.SHA256(),
        label=None
    )
)
```

```python
dados_recuperados = chave_privada_gamma.decrypt(
    dados_cifrados,
    padding.OAEP(
        mgf=padding.MGF1(algorithm=hashes.SHA256()),
        algorithm=hashes.SHA256(),
        label=None
    )
)
print(f"Dados: {dados_recuperados.decode()}")
```
